```python
import pandas as pd
```

```python
df = pd.read_csv("../data/raw/investments_VC.csv", encoding='latin-1')
df
```

```python
df = df.drop(columns=["permalink", 
                      "homepage_url", 
                      "post_ipo_equity",
                      "post_ipo_debt",
                      "last_funding_at",
                      "founded_year",
                      "founded_month",
                      "founded_quarter",
                      "category_list",
                      "state_code", 
                      "city"])
df
```

```python
print((df.isnull().mean() * 100).sort_values(ascending=False))
```

```python
print(df.dtypes)
```

```python
for c in ["founded_at", "first_funding_at"]:
    df[c] = pd.to_datetime(df[c], errors="coerce")
```

```python
df
```

```python
print(df.groupby(df["founded_at"].isna())["status"].value_counts(normalize=True))
```

```python
df = df.dropna(subset=["founded_at", "status"])
df
```

```python
df.columns = df.columns.str.strip()
```

```python
df["country_code"] = df["country_code"].fillna("Unknown")
df["region"] = df["region"].fillna("Unknown")
df["market"] = df["market"].fillna("Unknown")
```

```python
print((df.isnull().mean() * 100).sort_values(ascending=False))
```

```python
df = df.dropna(subset=["name"])
df
```

```python
print(df.dtypes)
```

```python
df["funding_total_usd"] = pd.to_numeric(
    df["funding_total_usd"].astype(str).str.replace(",", "").str.strip(),
    errors="coerce"
)

df
```

```python
print((df.isnull().mean() * 100).sort_values(ascending=False))
```

```python
df["funding_total_usd"] = df["funding_total_usd"].fillna(0)
df
```

```python
print(df.duplicated().sum())
```

```python
df[df.duplicated(subset=["name"], keep=False)].sort_values("name")
```

```python
df = df.sort_values("funding_total_usd", ascending=False)
df = df.drop_duplicates(subset=["name", "founded_at", "country_code"], keep="first")
```

```python
df
```

```python
print(df.isna().sum())
```

```python
print(df.dtypes)
```

```python
df["status"].unique()
```

```python
print(df["founded_at"].describe())
```

```python
print((df["founded_at"] < "1900-01-01").sum())
df[df["founded_at"] < "1900-01-01"][["name", "founded_at"]]
```

```python
df = df[df["founded_at"] >= "1900-01-01"]
```

```python
df = df.reset_index(drop=True)
df
```
