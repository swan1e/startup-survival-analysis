```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/proccessed/clean_investments_VC.csv")
df
```

```python
df.describe()
```

```python
df.describe(include="object")
```

```python
print(df["status"].value_counts(normalize=True) * 100)
```

```python
df["status"].value_counts().plot(kind="bar")
plt.title("Number of Startups by Status")
plt.xlabel("Status")
plt.ylabel("Number of Companies")
plt.show()
```

```python
df["country_code"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Countries by Number of Startups")
plt.xlabel("Country")
plt.ylabel("Number of Companies")
plt.show()
```

```python
df["market"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Markets by Number of Startups")
plt.xlabel("Market")
plt.ylabel("Number of Companies")
plt.show()
```

```python
pd.to_datetime(df["founded_at"], errors="coerce").dt.year.dropna().plot(
    kind="hist", bins=30
)
plt.title("Startups by Founding Year")
plt.xlabel("Founding Year")
plt.ylabel("Number of Companies")
plt.show()
```

```python
np.log1p(df["funding_total_usd"]).plot(kind="hist", bins=30)
plt.title("Distribution of Total Funding (log scale)")
plt.xlabel("log(1 + Funding in USD)")
plt.ylabel("Number of Companies")
plt.show()
```

```python
df["funding_rounds"].value_counts().sort_index().plot(kind="bar")
plt.title("Number of Startups by Funding Rounds")
plt.xlabel("Funding Rounds")
plt.ylabel("Number of Companies")
plt.show()
```

```python
print(df.groupby("status")[["funding_total_usd", "funding_rounds"]].median())
```

```python
top = df["country_code"].value_counts().head(10).index
print(df[df["country_code"].isin(top)].groupby("country_code")["status"].value_counts(normalize=True).unstack() * 100)
#country has little effect
```

```python
top = df["market"].value_counts().head(10).index
print(df[df["market"].isin(top)].groupby("market")["status"].value_counts(normalize=True).mul(100).unstack())
#market as a big effect, Curated Web(especially)
```

```python
top = df["founded_at"].value_counts().head(10).index
print(df[df["founded_at"].isin(top)].groupby("founded_at")["status"].value_counts(normalize=True).mul(100).unstack())
```

```python
top = df["funding_rounds"].value_counts().head(10).index
print(df.groupby("funding_rounds")["status"].value_counts(normalize=True).mul(100).unstack())
```
